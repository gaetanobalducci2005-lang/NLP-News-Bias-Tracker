```python
!pip install gnews
from gnews import GNews

def news_finder(query_user):

    filters = GNews(language="it", country="IT", period="7", max_results=10)

    news = filters.get_news(query_user)

    if len(news) == 0:
        print("Non ci sono articoli disponibili per questa richiesta, riprova.")
        return None
    else:
        return news
```

```python
#Simulazione inserimento dati utente
query_user = "Trump"

research_result = news_finder(query_user)

if research_result is not None:
    for article in research_result:
        print(article['title'])
        print(article['url'])
```
